```
using System;
using System.Collections.Generic;

class Notification
{
    private int _notificationId;
    private string _messageText;
    private string _type;
    private DateTime _createdAt;
    private bool _isRead;
    private string _senderName;

    public int NotificationId
    {
        get { return _notificationId; }
        set { _notificationId = value; }
    }

    public string MessageText
    {
        get { return _messageText; }
        set { _messageText = value; }
    }

    public string Type
    {
        get { return _type; }
        set { _type = value; }
    }

    public DateTime CreatedAt
    {
        get { return _createdAt; }
        set { _createdAt = value; }
    }

    public bool IsRead
    {
        get { return _isRead; }
        set { _isRead = value; }
    }

    public string SenderName
    {
        get { return _senderName; }
        set { _senderName = value; }
    }

    public Notification()
    {
        NotificationId = 0;
        MessageText = "Пустое уведомление";
        Type = "Base";
        CreatedAt = DateTime.Now;
        IsRead = false;
        SenderName = "Система";
    }

    public Notification(int id, string text, string type)
    {
        NotificationId = id;
        MessageText = text;
        Type = type;
        CreatedAt = DateTime.Now;
        IsRead = false;
        SenderName = "Система";
    }

    public Notification(int id, string text, string type, string sender)
    {
        NotificationId = id;
        MessageText = text;
        Type = type;
        SenderName = sender;
        CreatedAt = DateTime.Now;
        IsRead = false;
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine($"[{CreatedAt:HH:mm:ss}] ID: {NotificationId}, Тип: {Type}, От: {SenderName}, Прочитано: {(IsRead ? "да" : "нет")}");
        Console.WriteLine($"   Сообщение: {MessageText}");
    }

    public virtual void SendNotification()
    {
        Console.WriteLine($"Отправка {Type}...");
    }

    public virtual string GetNotificationDetails()
    {
        return $"ID: {NotificationId}, Тип: {Type}, Текст: {MessageText}, От: {SenderName}";
    }

    public virtual void ForwardTo(Notification other)
    {
        Console.WriteLine($"[{Type}] пересылает уведомление в [{other.Type}]...");
        other.MessageText = MessageText;
        other.SendNotification();
    }

    public void MarkAsRead()
    {
        IsRead = true;
        Console.WriteLine($"Уведомление {NotificationId} помечено как прочитанное.");
    }

    public void MarkAsRead(bool silent)
    {
        IsRead = true;
        if (!silent)
        {
            Console.WriteLine($"Уведомление {NotificationId} прочитано.");
        }
    }

    public virtual int GetUrgencyScore()
    {
        return 1;
    }

    public int GetUrgencyScore(int multiplier)
    {
        return GetUrgencyScore() * multiplier;
    }

    public virtual bool IsMoreImportantThan(Notification other)
    {
        return this.GetUrgencyScore() > other.GetUrgencyScore();
    }
}

interface IPriority
{
    int PriorityLevel { get; set; }
    void SetPriority(int level);
    string GetPriorityInfo();
}

interface ILoggable
{
    void Log(string message);
    string GetLogHistory();
}

interface ISchedulable
{
    DateTime ScheduledTime { get; set; }
    void Schedule(DateTime time);
    bool IsScheduled();
}

interface ITrackable
{
    string TrackingId { get; set; }
    void Track();
}

class EmailNotification : Notification, IPriority, ILoggable
{
    private string _emailAddress;
    private string _subject;
    private int _priorityLevel;
    private string _logHistory;
    private string _attachmentName;
    private bool _isEncrypted;

    public string EmailAddress
    {
        get { return _emailAddress; }
        set { _emailAddress = value; }
    }

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

    public int PriorityLevel
    {
        get { return _priorityLevel; }
        set { _priorityLevel = value; }
    }

    public string AttachmentName
    {
        get { return _attachmentName; }
        set { _attachmentName = value; }
    }

    public bool IsEncrypted
    {
        get { return _isEncrypted; }
        set { _isEncrypted = value; }
    }

    public EmailNotification() : base()
    {
        Type = "Email";
        EmailAddress = "unknown@mail.ru";
        Subject = "Без темы";
        PriorityLevel = 1;
        _logHistory = "";
        AttachmentName = "нет";
        IsEncrypted = false;
    }

    public EmailNotification(int id, string text, string email) : base(id, text, "Email")
    {
        EmailAddress = email;
        Subject = "Без темы";
        PriorityLevel = 1;
        _logHistory = "";
        AttachmentName = "нет";
        IsEncrypted = false;
    }

    public EmailNotification(int id, string text, string email, string subject, int priority) : base(id, text, "Email")
    {
        EmailAddress = email;
        Subject = subject;
        PriorityLevel = priority;
        _logHistory = "";
        AttachmentName = "нет";
        IsEncrypted = false;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка Email на {EmailAddress}: {MessageText}");
        if (IsEncrypted)
        {
            Console.WriteLine("   (письмо зашифровано)");
        }
        Log($"Email отправлен на {EmailAddress}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Email: {EmailAddress}, Тема: {Subject}, Приоритет: {PriorityLevel}, Вложение: {AttachmentName}";
    }

    public void SetPriority(int level)
    {
        PriorityLevel = level;
        Console.WriteLine($"Приоритет Email установлен на {level}");
    }

    public string GetPriorityInfo()
    {
        return $"Email приоритет: {PriorityLevel}";
    }

    public void Log(string message)
    {
        _logHistory += $"[{DateTime.Now}] {message}\n";
    }

    public string GetLogHistory()
    {
        return _logHistory;
    }

    public override int GetUrgencyScore()
    {
        return PriorityLevel * 2;
    }

    public void AttachFile(string fileName)
    {
        AttachmentName = fileName;
        Console.WriteLine($"К письму прикреплён файл: {fileName}");
    }

    public void AttachFile(string fileName, bool encrypt)
    {
        AttachmentName = fileName;
        IsEncrypted = encrypt;
        Console.WriteLine($"К письму прикреплён файл: {fileName} (шифрование: {(encrypt ? "да" : "нет")})");
    }
}

class SMSNotification : Notification, IPriority
{
    private string _phoneNumber;
    private int _priorityLevel;
    private string _operator;
    private bool _isDeliveryReport;
    private int _characterCount;

    public string PhoneNumber
    {
        get { return _phoneNumber; }
        set { _phoneNumber = value; }
    }

    public int PriorityLevel
    {
        get { return _priorityLevel; }
        set { _priorityLevel = value; }
    }

    public string Operator
    {
        get { return _operator; }
        set { _operator = value; }
    }

    public bool IsDeliveryReport
    {
        get { return _isDeliveryReport; }
        set { _isDeliveryReport = value; }
    }

    public int CharacterCount
    {
        get { return _characterCount; }
        set { _characterCount = value; }
    }

    public SMSNotification() : base()
    {
        Type = "SMS";
        PhoneNumber = "+0-000-000-00-00";
        PriorityLevel = 1;
        Operator = "Unknown";
        IsDeliveryReport = false;
        CharacterCount = 0;
    }

    public SMSNotification(int id, string text, string phone) : base(id, text, "SMS")
    {
        PhoneNumber = phone;
        PriorityLevel = 1;
        Operator = "Unknown";
        IsDeliveryReport = false;
        CharacterCount = text.Length;
    }

    public SMSNotification(int id, string text, string phone, string oper, int priority) : base(id, text, "SMS")
    {
        PhoneNumber = phone;
        Operator = oper;
        PriorityLevel = priority;
        IsDeliveryReport = false;
        CharacterCount = text.Length;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка SMS на {PhoneNumber} ({Operator}): {MessageText}");
        Console.WriteLine($"   Символов: {CharacterCount}, Отчёт о доставке: {(IsDeliveryReport ? "вкл" : "выкл")}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Телефон: {PhoneNumber}, Оператор: {Operator}, Приоритет: {PriorityLevel}";
    }

    public void SetPriority(int level)
    {
        PriorityLevel = level;
        Console.WriteLine($"Приоритет SMS установлен на {level}");
    }

    public string GetPriorityInfo()
    {
        return $"SMS приоритет: {PriorityLevel}";
    }

    public override int GetUrgencyScore()
    {
        return PriorityLevel + 1;
    }

    public void EnableDeliveryReport(bool enable)
    {
        IsDeliveryReport = enable;
        Console.WriteLine($"Отчёт о доставке SMS: {(enable ? "включён" : "выключен")}");
    }
}

class PushNotification : Notification, IPriority, ILoggable, ISchedulable, ITrackable
{
    private string _platform;
    private string _video;
    private string _appName;
    private int _priorityLevel;
    private string _logHistory;
    private DateTime _scheduledTime;
    private string _trackingId;
    private bool _isSilent;

    public string Platform
    {
        get { return _platform; }
        set { _platform = value; }
    }

    public string Video
    {
        get { return _video; }
        set { _video = value; }
    }

    public string AppName
    {
        get { return _appName; }
        set { _appName = value; }
    }

    public int PriorityLevel
    {
        get { return _priorityLevel; }
        set { _priorityLevel = value; }
    }

    public DateTime ScheduledTime
    {
        get { return _scheduledTime; }
        set { _scheduledTime = value; }
    }

    public string TrackingId
    {
        get { return _trackingId; }
        set { _trackingId = value; }
    }

    public bool IsSilent
    {
        get { return _isSilent; }
        set { _isSilent = value; }
    }

    public PushNotification() : base()
    {
        Type = "Push";
        Platform = "Unknown";
        Video = "none.mp4";
        AppName = "UnknownApp";
        PriorityLevel = 1;
        _logHistory = "";
        _scheduledTime = DateTime.MinValue;
        _trackingId = Guid.NewGuid().ToString().Substring(0, 8);
        _isSilent = false;
    }

    public PushNotification(int id, string text, string platform, string video, string appName)
        : base(id, text, "Push")
    {
        Platform = platform;
        Video = video;
        AppName = appName;
        PriorityLevel = 1;
        _logHistory = "";
        _scheduledTime = DateTime.MinValue;
        _trackingId = Guid.NewGuid().ToString().Substring(0, 8);
        _isSilent = false;
    }

    public PushNotification(int id, string text, string platform, string video, string appName, int priority)
        : base(id, text, "Push")
    {
        Platform = platform;
        Video = video;
        AppName = appName;
        PriorityLevel = priority;
        _logHistory = "";
        _scheduledTime = DateTime.MinValue;
        _trackingId = Guid.NewGuid().ToString().Substring(0, 8);
        _isSilent = false;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка Push на платформу {Platform} (приложение {AppName}): {MessageText}");
        Console.WriteLine($"   Tracking ID: {TrackingId}, Тихий режим: {(IsSilent ? "да" : "нет")}");
        Log($"Push отправлен на {Platform}");
    }

    public override void DisplayNotification()
    {
        Console.WriteLine($"[Мобильное уведомление] Платформа: {Platform}, Приложение: {AppName}, Видео: {Video}");
        Console.WriteLine($"   Сообщение: {MessageText}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Платформа: {Platform}, Видео: {Video}, Приложение: {AppName}, Приоритет: {PriorityLevel}";
    }

    public void SetPriority(int level)
    {
        PriorityLevel = level;
        Console.WriteLine($"Приоритет Push установлен на {level}");
    }

    public string GetPriorityInfo()
    {
        return $"Push приоритет: {PriorityLevel}";
    }

    public void Log(string message)
    {
        _logHistory += $"[{DateTime.Now}] {message}\n";
    }

    public string GetLogHistory()
    {
        return _logHistory;
    }

    public void Schedule(DateTime time)
    {
        ScheduledTime = time;
        Console.WriteLine($"Push запланирован на {time}");
    }

    public bool IsScheduled()
    {
        return ScheduledTime > DateTime.MinValue;
    }

    public void Track()
    {
        Console.WriteLine($"Отслеживание Push по ID: {TrackingId}");
    }

    public override int GetUrgencyScore()
    {
        return PriorityLevel * 3;
    }
}

class UrgentNotification : EmailNotification
{
    private int _urgencyLevel;
    private string _reason;

    public int UrgencyLevel
    {
        get { return _urgencyLevel; }
        set { _urgencyLevel = value; }
    }

    public string Reason
    {
        get { return _reason; }
        set { _reason = value; }
    }

    public UrgentNotification() : base()
    {
        Type = "UrgentEmail";
        UrgencyLevel = 5;
        Reason = "Не указана";
    }

    public UrgentNotification(int id, string text, string email, string subject, int priority, int urgency, string reason)
        : base(id, text, email, subject, priority)
    {
        Type = "UrgentEmail";
        UrgencyLevel = urgency;
        Reason = reason;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"[СРОЧНО!] Отправка Email на {EmailAddress}: {MessageText}");
        Console.WriteLine($"   Причина срочности: {Reason}, Уровень: {UrgencyLevel}");
        Log($"Срочный Email отправлен на {EmailAddress}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Срочность: {UrgencyLevel}, Причина: {Reason}";
    }

    public override int GetUrgencyScore()
    {
        return base.GetUrgencyScore() + UrgencyLevel * 10;
    }

    public void Escalate()
    {
        UrgencyLevel++;
        Console.WriteLine($"Уровень срочности повышен до {UrgencyLevel}");
    }

    public void Escalate(int levels)
    {
        UrgencyLevel += levels;
        Console.WriteLine($"Уровень срочности повышен на {levels} до {UrgencyLevel}");
    }
}

class NotificationCollection<T> where T : Notification
{
    private List<T> _items = new List<T>();

    public void Add(T item)
    {
        _items.Add(item);
        Console.WriteLine($"Добавлено уведомление типа {typeof(T).Name} (ID: {item.NotificationId})");
    }

    public void Remove(T item)
    {
        _items.Remove(item);
    }

    public int Count
    {
        get { return _items.Count; }
    }

    public void DisplayAll()
    {
        Console.WriteLine($"=== Коллекция {typeof(T).Name} ({Count} шт.) ===");
        foreach (var item in _items)
        {
            item.DisplayNotification();
            Console.WriteLine();
        }
    }

    public T FindMostUrgent()
    {
        if (_items.Count == 0) return null;

        T mostUrgent = _items[0];
        foreach (var item in _items)
        {
            if (item.GetUrgencyScore() > mostUrgent.GetUrgencyScore())
            {
                mostUrgent = item;
            }
        }
        return mostUrgent;
    }

    public List<T> FilterBy(Func<T, bool> predicate)
    {
        List<T> result = new List<T>();
        foreach (var item in _items)
        {
            if (predicate(item))
            {
                result.Add(item);
            }
        }
        return result;
    }
}

class NotificationService
{
    public static void SendAll<T>(IEnumerable<T> notifications) where T : Notification
    {
        Console.WriteLine($"--- Массовая отправка ({typeof(T).Name}) ---");
        foreach (var n in notifications)
        {
            n.SendNotification();
        }
    }

    public static T FindMaxUrgency<T>(IEnumerable<T> notifications) where T : Notification
    {
        T max = null;
        foreach (var n in notifications)
        {
            if (max == null || n.GetUrgencyScore() > max.GetUrgencyScore())
            {
                max = n;
            }
        }
        return max;
    }
}

        Notification[] notifications = new Notification[]
        {
            new EmailNotification(1, "Письмо", "a@mail.ru", "Тема", 2),
            new SMSNotification(2, "Код", "+7-999-123-45-67", "MTS", 3),
            new PushNotification(3, "Пуш", "iOS", "v.mp4", "App", 4),
            new UrgentNotification(4, "Срочно", "b@mail.ru", "Важно", 5, 5, "Дедлайн")
        };

        foreach (Notification n in notifications)
        {
            n.DisplayNotification();
            n.SendNotification();
            Console.WriteLine($"   Детали: {n.GetNotificationDetails()}");
            Console.WriteLine($"   Оценка срочности: {n.GetUrgencyScore()}");
            Console.WriteLine();
        }

        EmailNotification email = new EmailNotification(10, "Тест", "test@mail.ru", "Тема", 2);
        email.AttachFile("doc.pdf");
        email.AttachFile("secret.pdf", true);
        email.MarkAsRead();
        email.MarkAsRead(true);

        UrgentNotification urgent = new UrgentNotification(11, "Срочно!", "boss@mail.ru", "Важно", 5, 5, "Дедлайн");
        urgent.Escalate();
        urgent.Escalate(3);

        Console.WriteLine();

        NotificationCollection<EmailNotification> emailBox = new NotificationCollection<EmailNotification>();
        emailBox.Add(new EmailNotification(20, "Письмо 1", "a@mail.ru", "Тема 1", 1));
        emailBox.Add(new EmailNotification(21, "Письмо 2", "b@mail.ru", "Тема 2", 5));
        emailBox.Add(new EmailNotification(22, "Письмо 3", "c@mail.ru", "Тема 3", 3));
        emailBox.DisplayAll();

        var mostUrgent = emailBox.FindMostUrgent();
        Console.WriteLine($"Самое срочное письмо: ID {mostUrgent.NotificationId}\n");

        NotificationCollection<Notification> allBox = new NotificationCollection<Notification>();
        allBox.Add(new EmailNotification(30, "Email", "x@mail.ru"));
        allBox.Add(new SMSNotification(31, "SMS", "+7-000-000-00-00"));
        allBox.Add(new PushNotification(32, "Push", "Android", "v.mp4", "App"));
        allBox.Add(new UrgentNotification(33, "Urgent", "y@mail.ru", "SOS", 5, 9, "Авария"));

        allBox.DisplayAll();
        Console.WriteLine($"Самое срочное: ID {allBox.FindMostUrgent().NotificationId}\n");

        NotificationService.SendAll(allBox.FilterBy(n => n.GetUrgencyScore() > 3));
        Console.WriteLine();

        var maxUrgency = NotificationService.FindMaxUrgency(allBox.FilterBy(n => true));
        Console.WriteLine($"Максимальная срочность: {maxUrgency.GetNotificationDetails()}");
```
