```
using System;

class Notification
{
    private int _notificationId;
    private string _messageText;
    private string _type;

    public int NotificationId
    {
        get { return _notificationId; }
        set { _notificationId = value; }
    }

    public string MessageText
    {
        get { return _messageText; }
        set { _messageText = value; }
    }

    public string Type
    {
        get { return _type; }
        set { _type = value; }
    }

    public Notification()
    {
        NotificationId = 0;
        MessageText = "Пустое уведомление";
        Type = "Base";
    }

    public Notification(int id, string text, string type)
    {
        NotificationId = id;
        MessageText = text;
        Type = type;
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine($"ID: {NotificationId}, Тип: {Type}, Сообщение: {MessageText}");
    }

    public virtual void SendNotification()
    {
        Console.WriteLine($"Отправка {Type}...");
    }

    public virtual string GetNotificationDetails()
    {
        return $"ID: {NotificationId}, Тип: {Type}, Текст: {MessageText}";
    }

    public virtual void ForwardTo(Notification other)
    {
        Console.WriteLine($"[{Type}] пересылает уведомление в [{other.Type}]...");
        other.MessageText = MessageText;
        other.SendNotification();
    }
}

interface IPriority
{
    int PriorityLevel { get; set; }
    void SetPriority(int level);
    string GetPriorityInfo();
}

interface ILoggable
{
    void Log(string message);
    string GetLogHistory();
}

interface ISchedulable
{
    DateTime ScheduledTime { get; set; }
    void Schedule(DateTime time);
    bool IsScheduled();
}

class EmailNotification : Notification, IPriority, ILoggable
{
    private string _emailAddress;
    private string _subject;
    private int _priorityLevel;
    private string _logHistory;

    public string EmailAddress
    {
        get { return _emailAddress; }
        set { _emailAddress = value; }
    }

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

    public int PriorityLevel
    {
        get { return _priorityLevel; }
        set { _priorityLevel = value; }
    }

    public EmailNotification() : base()
    {
        Type = "Email";
        EmailAddress = "unknown@mail.ru";
        Subject = "Без темы";
        PriorityLevel = 1;
        _logHistory = "";
    }

    public EmailNotification(int id, string text, string email) : base(id, text, "Email")
    {
        EmailAddress = email;
        Subject = "Без темы";
        PriorityLevel = 1;
        _logHistory = "";
    }

    public EmailNotification(int id, string text, string email, string subject, int priority) : base(id, text, "Email")
    {
        EmailAddress = email;
        Subject = subject;
        PriorityLevel = priority;
        _logHistory = "";
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка Email на {EmailAddress}: {MessageText}");
        Log($"Email отправлен на {EmailAddress}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Email: {EmailAddress}, Тема: {Subject}, Приоритет: {PriorityLevel}";
    }

    public void SetPriority(int level)
    {
        PriorityLevel = level;
        Console.WriteLine($"Приоритет Email установлен на {level}");
    }

    public string GetPriorityInfo()
    {
        return $"Email приоритет: {PriorityLevel}";
    }

    public void Log(string message)
    {
        _logHistory += $"[{DateTime.Now}] {message}\n";
    }

    public string GetLogHistory()
    {
        return _logHistory;
    }
}

class SMSNotification : Notification, IPriority
{
    private string _phoneNumber;
    private int _priorityLevel;
    private string _operator;

    public string PhoneNumber
    {
        get { return _phoneNumber; }
        set { _phoneNumber = value; }
    }

    public int PriorityLevel
    {
        get { return _priorityLevel; }
        set { _priorityLevel = value; }
    }

    public string Operator
    {
        get { return _operator; }
        set { _operator = value; }
    }

    public SMSNotification() : base()
    {
        Type = "SMS";
        PhoneNumber = "+0-000-000-00-00";
        PriorityLevel = 1;
        Operator = "Unknown";
    }

    public SMSNotification(int id, string text, string phone) : base(id, text, "SMS")
    {
        PhoneNumber = phone;
        PriorityLevel = 1;
        Operator = "Unknown";
    }

    public SMSNotification(int id, string text, string phone, string oper, int priority) : base(id, text, "SMS")
    {
        PhoneNumber = phone;
        Operator = oper;
        PriorityLevel = priority;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка SMS на {PhoneNumber}: {MessageText}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Телефон: {PhoneNumber}, Оператор: {Operator}, Приоритет: {PriorityLevel}";
    }

    public void SetPriority(int level)
    {
        PriorityLevel = level;
        Console.WriteLine($"Приоритет SMS установлен на {level}");
    }

    public string GetPriorityInfo()
    {
        return $"SMS приоритет: {PriorityLevel}";
    }
}

class PushNotification : Notification, IPriority, ILoggable, ISchedulable
{
    private string _platform;
    private string _video;
    private string _appName;
    private int _priorityLevel;
    private string _logHistory;
    private DateTime _scheduledTime;

    public string Platform
    {
        get { return _platform; }
        set { _platform = value; }
    }

    public string Video
    {
        get { return _video; }
        set { _video = value; }
    }

    public string AppName
    {
        get { return _appName; }
        set { _appName = value; }
    }

    public int PriorityLevel
    {
        get { return _priorityLevel; }
        set { _priorityLevel = value; }
    }

    public DateTime ScheduledTime
    {
        get { return _scheduledTime; }
        set { _scheduledTime = value; }
    }

    public PushNotification() : base()
    {
        Type = "Push";
        Platform = "Unknown";
        Video = "none.mp4";
        AppName = "UnknownApp";
        PriorityLevel = 1;
        _logHistory = "";
        _scheduledTime = DateTime.MinValue;
    }

    public PushNotification(int id, string text, string platform, string video, string appName)
        : base(id, text, "Push")
    {
        Platform = platform;
        Video = video;
        AppName = appName;
        PriorityLevel = 1;
        _logHistory = "";
        _scheduledTime = DateTime.MinValue;
    }

    public PushNotification(int id, string text, string platform, string video, string appName, int priority)
        : base(id, text, "Push")
    {
        Platform = platform;
        Video = video;
        AppName = appName;
        PriorityLevel = priority;
        _logHistory = "";
        _scheduledTime = DateTime.MinValue;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправка Push на платформу {Platform} (приложение {AppName}): {MessageText}");
        Log($"Push отправлен на {Platform}");
    }

    public override void DisplayNotification()
    {
        Console.WriteLine($"[Мобильное уведомление] Платформа: {Platform}, Приложение: {AppName}, Видео: {Video}, Сообщение: {MessageText}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Платформа: {Platform}, Видео: {Video}, Приложение: {AppName}, Приоритет: {PriorityLevel}";
    }

    public void SetPriority(int level)
    {
        PriorityLevel = level;
        Console.WriteLine($"Приоритет Push установлен на {level}");
    }

    public string GetPriorityInfo()
    {
        return $"Push приоритет: {PriorityLevel}";
    }

    public void Log(string message)
    {
        _logHistory += $"[{DateTime.Now}] {message}\n";
    }

    public string GetLogHistory()
    {
        return _logHistory;
    }

    public void Schedule(DateTime time)
    {
        ScheduledTime = time;
        Console.WriteLine($"Push запланирован на {time}");
    }

    public bool IsScheduled()
    {
        return ScheduledTime > DateTime.MinValue;
    }
}

class UrgentNotification : EmailNotification
{
    private int _urgencyLevel;
    private string _reason;

    public int UrgencyLevel
    {
        get { return _urgencyLevel; }
        set { _urgencyLevel = value; }
    }

    public string Reason
    {
        get { return _reason; }
        set { _reason = value; }
    }

    public UrgentNotification() : base()
    {
        Type = "UrgentEmail";
        UrgencyLevel = 5;
        Reason = "Не указана";
    }

    public UrgentNotification(int id, string text, string email, string subject, int priority, int urgency, string reason)
        : base(id, text, email, subject, priority)
    {
        Type = "UrgentEmail";
        UrgencyLevel = urgency;
        Reason = reason;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"[СРОЧНО!] Отправка Email на {EmailAddress}: {MessageText}");
        Console.WriteLine($"Причина срочности: {Reason}, Уровень: {UrgencyLevel}");
        Log($"Срочный Email отправлен на {EmailAddress}");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Срочность: {UrgencyLevel}, Причина: {Reason}";
    }

    public void Escalate()
    {
        UrgencyLevel++;
        Console.WriteLine($"Уровень срочности повышен до {UrgencyLevel}");
    }
}

        Console.WriteLine("Простое наследование");
        EmailNotification email = new EmailNotification(1, "Привет!", "user@mail.ru", "Тест", 2);
        email.SendNotification();
        Console.WriteLine(email.GetNotificationDetails());
        email.SetPriority(3);
        Console.WriteLine(email.GetPriorityInfo());
        email.Log("Тест лога");
        Console.WriteLine(email.GetLogHistory());

        Console.WriteLine("\nИерархическое наследование (UrgentNotification -> EmailNotification)");
        UrgentNotification urgent = new UrgentNotification(2, "Срочно!", "boss@mail.ru", "Важно", 5, 5, "Дедлайн");
        urgent.SendNotification();
        Console.WriteLine(urgent.GetNotificationDetails());
        urgent.Escalate();
        Console.WriteLine(urgent.GetNotificationDetails());

        Console.WriteLine("\nМножественное наследование через интерфейсы");
        PushNotification push = new PushNotification(3, "Новое видео!", "Android", "video.mp4", "YouTube", 3);
        push.SendNotification();
        push.Log("Push лог");
        Console.WriteLine(push.GetLogHistory());
        push.Schedule(DateTime.Now.AddHours(2));
        Console.WriteLine($"Запланировано: {push.IsScheduled()}");

        Console.WriteLine("\nПолиморфизм");
        Notification[] notifications = new Notification[]
        {
            new EmailNotification(4, "Письмо", "a@mail.ru"),
            new SMSNotification(5, "Код", "+7-999-123-45-67"),
            new PushNotification(6, "Пуш", "iOS", "v.mp4", "App"),
            new UrgentNotification(7, "Срочно", "b@mail.ru", "Тема", 5, 5, "Причина")
        };

        foreach (Notification notif in notifications)
        {
            notif.DisplayNotification();
            notif.SendNotification();
            Console.WriteLine(notif.GetNotificationDetails());

            if (notif is IPriority priority)
            {
                Console.WriteLine(priority.GetPriorityInfo());
            }

            if (notif is ILoggable loggable)
            {
                Console.WriteLine($"Лог: {loggable.GetLogHistory()}");
            }

            Console.WriteLine();
        }

        Console.WriteLine("Взаимодействие объектов");
        Notification email2 = new EmailNotification(8, "Важное сообщение", "boss@mail.ru");
        Notification sms = new SMSNotification(9, "Пусто", "+7-000-000-00-00");
        email2.ForwardTo(sms);
        sms.DisplayNotification();
```
