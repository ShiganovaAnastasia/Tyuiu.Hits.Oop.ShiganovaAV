Разработать систему уведомлений на основе принципов ООП. Базовый класс Notification определяет общую структуру уведомлений. Производные классы EmailNotification, SMSNotification и PushNotification расширяют функциональность для различных каналов доставки.

Дополнительное задание

```
using System;

public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;
//конструктор для создания экземпляра
    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }
//методы геттеры(читать) и сеттеры(изменять) для приватных полей
    public string Name { get { return name; } set { name = value; } }
    public decimal Price { get { return price; } set { price = value; } }
    public int Quantity { get { return quantity; } set { quantity = value; } }
//возвращает общую стоим-ть всех ед товар
    public decimal GetTotalPrice()
    {
        return price * quantity;
    }
//обновляет количество товара на указанное знач-e
    public void UpdateQuantity(int amount)
    {
        quantity += amount;
    }
//возвращает строку с полным описанием товара
    public string GetProductInfo()
    {
        return $"{name} ({category}) - {price} руб. Кол-во: {quantity}. шт. Описание: {description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
//допсписок для хранения товаров продавца
    private List<Product> products = new List<Product>();

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
    }

    public string Name { get { return name; } set { name = value; } }
//добавляет товар в список товаров продавца
    public void AddProduct(Product product)
    {
        products.Add(product);
    }
//уменьшает количество товара на складе
    public void SellProduct(Product product, int quantity)
    {
        product.UpdateQuantity(-quantity);
    }
//возвращает информацию о продавце
    public string GetSellerInfo()
    {
        return $"Продавец: {name}, ID: {employeeId}, Контакт: {contactInfo}, Зарплата: {salary} руб.";
    }

    public List<Product> GetProducts()
    {
        return products;
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
//cписок продавцов магазина
    private List<Seller> sellers = new List<Seller>();

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }
//добавляет продавца в магазин
    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
    }
//перечисляет все товары в магазине
    public void ListProducts()
    {
        foreach (var seller in sellers)
        {
            foreach (var product in seller.GetProducts())
            {
                Console.WriteLine(product.GetProductInfo());
            }
        }
    }
//возвращает информацию о магазине
    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}, Адрес: {location}, Время работы: {storeHours}";
    }
}
//cоздание по одному экземпляру каждого класса
Product product = new Product("Принтер", 10000m, 5, "Принтеры", "Лазерный принтер");
Seller seller = new Seller("Арина", "568930", 60000m, "arinara@mail.ru");
Store store = new Store("ДНС", "ул. Республики 65", "09:00-21:00");
store.AddSeller(seller);
seller.AddProduct(product);
Console.WriteLine(store.GetStoreInfo());
Console.WriteLine(seller.GetSellerInfo());
Console.WriteLine(product.GetProductInfo()); 
Console.WriteLine($"Общая стоимость: {product.GetTotalPrice()} руб.");
seller.SellProduct(product, 3);
Console.WriteLine($"После продажи: {product.GetProductInfo()}");
Console.WriteLine("Товары в магазине:");
store.ListProducts();
```

```
using System;

// БАЗОВЫЙ КЛАСС Notification — общая структура для всех уведомлений.
class Notification
{
    // ПРИВАТНЫЕ ПОЛЯ — спрятаны, работаем через свойства.
    private int _notificationId;
    private string _messageText;
    private string _type;

    // СВОЙСТВО NotificationId — доступ к _notificationId через get/set.
    public int NotificationId
    {
        get { return _notificationId; }
        set { _notificationId = value; }
    }

    // СВОЙСТВО MessageText — доступ к _messageText.
    public string MessageText
    {
        get { return _messageText; }
        set { _messageText = value; }
    }

    // СВОЙСТВО Type — доступ к _type (Email/SMS/Push).
    public string Type
    {
        get { return _type; }
        set { _type = value; }
    }

    // КОНСТРУКТОР ПО УМОЛЧАНИЮ — срабатывает при new Notification().
    public Notification()
    {
        NotificationId = 0;
        MessageText = "Пустое уведомление";
        Type = "Base";
    }

    // КОНСТРУКТОР С ПАРАМЕТРАМИ — задаём все поля сразу.
    public Notification(int id, string text, string type)
    {
        NotificationId = id;
        MessageText = text;
        Type = type;
    }

    // VIRTUAL — метод можно ПЕРЕОПРЕДЕЛИТЬ в наследниках.
    // Показывает уведомление в консоль.
    public virtual void DisplayNotification()
    {
        Console.WriteLine($"ID: {NotificationId}, Тип: {Type}, Сообщение: {MessageText}");
    }

    // VIRTUAL — можно переопределить. Просто отправка.
    public virtual void SendNotification()
    {
        Console.WriteLine($"Отправка {Type}...");
    }

    // VIRTUAL — можно переопределить. Возвращает строку с деталями.
    public virtual string GetNotificationDetails()
    {
        return $"ID: {NotificationId}, Тип: {Type}, Текст: {MessageText}";
    }

    // ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ — принимает ДРУГОЕ уведомление (other).
    // Копирует свой текст в другое уведомление и отправляет его.
    public virtual void ForwardTo(Notification other)
    {
        Console.WriteLine($"[{Type}] пересылает уведомление в [{other.Type}]...");
        other.MessageText = MessageText;   // копируем текст в другое уведомление
        other.SendNotification();          // вызываем метод у ДРУГОГО объекта
    }
}

// НАСЛЕДНИК EmailNotification — уведомление на почту.
class EmailNotification : Notification
{
    // СВОЁ ПОЛЕ — адрес почты.
    private string _emailAddress;

    // СВОЙСТВО EmailAddress — доступ к _emailAddress.
    public string EmailAddress
    {
        get { return _emailAddress; }
        set { _emailAddress = value; }
    }

    // КОНСТРУКТОР ПО УМОЛЧАНИЮ. : base() — сначала Notification().
    public EmailNotification() : base()
    {
        Type = "Email";                          // свой тип
        EmailAddress = "unknown@mail.ru";        // значение по умолчанию
    }

    // КОНСТРУКТОР С ПАРАМЕТРАМИ. : base(id, text, "Email") — родитель настроит ID, текст, тип.
    public EmailNotification(int id, string text, string email) : base(id, text, "Email")
    {
        EmailAddress = email;   // своя настройка — адрес
    }

    // OVERRIDE — ЗАМЕНЯЕМ версию родителя SendNotification.
    public override void SendNotification()
    {
        Console.WriteLine($"Отправка Email на {EmailAddress}: {MessageText}");
    }

    // OVERRIDE — ЗАМЕНЯЕМ GetNotificationDetails. Вызываем base.GetNotificationDetails() — версию родителя, потом добавляем своё.
    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Email: {EmailAddress}";
        //      берём базовые детали        добавляем адрес Email
    }
}

// НАСЛЕДНИК SMSNotification — SMS-уведомление.
class SMSNotification : Notification
{
    // СВОЁ ПОЛЕ — номер телефона.
    private string _phoneNumber;

    // СВОЙСТВО PhoneNumber — доступ к _phoneNumber.
    public string PhoneNumber
    {
        get { return _phoneNumber; }
        set { _phoneNumber = value; }
    }

    // КОНСТРУКТОР ПО УМОЛЧАНИЮ.
    public SMSNotification() : base()
    {
        Type = "SMS";
        PhoneNumber = "+0-000-000-00-00";
    }

    // КОНСТРУКТОР С ПАРАМЕТРАМИ.
    public SMSNotification(int id, string text, string phone) : base(id, text, "SMS")
    {
        PhoneNumber = phone;
    }

    // OVERRIDE — своя отправка SMS.
    public override void SendNotification()
    {
        Console.WriteLine($"Отправка SMS на {PhoneNumber}: {MessageText}");
    }

    // OVERRIDE — свои детали + телефон.
    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Телефон: {PhoneNumber}";
    }
}

// НАСЛЕДНИК PushNotification — пуш-уведомление в приложение.
class PushNotification : Notification
{
    // СВОИ ПОЛЯ — платформа, видео, имя приложения.
    private string _platform;
    private string _video;
    private string _appName;

    public string Platform
    {
        get { return _platform; }
        set { _platform = value; }
    }

    public string Video
    {
        get { return _video; }
        set { _video = value; }
    }

    public string AppName
    {
        get { return _appName; }
        set { _appName = value; }
    }

    // КОНСТРУКТОР ПО УМОЛЧАНИЮ.
    public PushNotification() : base()
    {
        Type = "Push";
        Platform = "Unknown";
        Video = "none.mp4";
        AppName = "UnknownApp";
    }

    // КОНСТРУКТОР С ПАРАМЕТРАМИ — 5 значений.
    public PushNotification(int id, string text, string platform, string video, string appName)
        : base(id, text, "Push")
    {
        Platform = platform;
        Video = video;
        AppName = appName;
    }

    // OVERRIDE — своя отправка Push.
    public override void SendNotification()
    {
        Console.WriteLine($"Отправка Push на платформу {Platform} (приложение {AppName}): {MessageText}");
    }

    // OVERRIDE — СВОЯ версия показа уведомления (только у Push).
    public override void DisplayNotification()
    {
        Console.WriteLine($"[Мобильное уведомление] Платформа: {Platform}, Приложение: {AppName}, Видео: {Video}, Сообщение: {MessageText}");
    }

    // OVERRIDE — свои детали: платформа, видео, приложение.
    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Платформа: {Platform}, Видео: {Video}, Приложение: {AppName}";
    }
}

//Запуск

// МАССИВ типа Notification — может хранить ЛЮБЫХ наследников (полиморфизм).
Notification[] notifications = new Notification[]
{
    new EmailNotification(1, "Привет!", "user@mail.ru"),
    new SMSNotification(2, "Код: 1234", "+7-999-123-45-67"),
    new PushNotification(3, "Новое видео!", "Android", "video_123.mp4", "YouTube")
};

// ЦИКЛ — проходим по всем уведомлениям.
// Благодаря override вызываются СВОИ версии методов у каждого наследника.
foreach (Notification notif in notifications)
{
    notif.DisplayNotification();                      // у Push — своя версия
    notif.SendNotification();                         // у каждого своя
    Console.WriteLine(notif.GetNotificationDetails()); // у каждого своя
    Console.WriteLine();                              // пустая строка для читаемости
}

//Взаимодействие объектов
// Создаём два уведомления разных типов
Notification email = new EmailNotification(4, "Важное сообщение", "boss@mail.ru");
Notification sms = new SMSNotification(5, "Пусто", "+7-000-000-00-00");

// email передаёт своё сообщение в sms и отправляет его
email.ForwardTo(sms);

// Проверяем, что текст скопировался
sms.DisplayNotification();
// "ID: 5, Тип: SMS, Сообщение: Важное сообщение"
//   MessageText был "Пусто", стал "Важное сообщение" — ForwardTo скопировал
```
